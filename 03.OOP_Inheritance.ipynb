Inheritance

Inheritance ka matlab hai ek class doosri class ke attributes aur methods ko inherit/reuse kar sakti hai.

Simple example:

Animal ek parent class hai.
Dog ek child class hai.
Dog, Animal ki properties/methods use kar sakta hai.

```python
class Animal:
    def eat(self):
        print("Animal is eating")


class Dog(Animal):
    pass


dog1 = Dog()

dog1.eat()
```

```python
class Vehicle:
    def move(self):
        print("Vehicle is moving")

class Car(Vehicle):
    pass
car1=Car()
car1.move()
```

```python
class Vehicle:
    def move(self):
        print("Vehicle is moving")


class Car(Vehicle):
    def horn(self):
        print("Car is honking")


car1 = Car()

car1.move()
car1.horn()
```

```python
class Animal():
    def move(self):
        print("Animal is eating")
class Dog(Animal):
    def eat(self):
       print("dog is barking")
Bark=Dog()
Bark.move()
Bark.eat()
```

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        print(self.name, "is eating")


class Dog(Animal):
    pass


dog1 = Dog("Tommy")

print(dog1.name)
dog1.eat()
```

```python
class Person:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        print(self.name, "is my name")


class Student(Person):
    pass


student1 = Student("Abdullah")

student1.introduce()
```

Inheritance — super()

super() ka use child class se parent class ke constructor ya method ko call karne ke liye hota hai

```python
class Person:
    def __init__(self, name):
        self.name = name


class Student(Person):
    def __init__(self, name, age):
        super().__init__(name)
        self.age = age


student1 = Student("Abdullah", 20)

print(student1.name)
print(student1.age)
```

```python
class Person:
     def __init__(self, name):
        self.name=name
class Student(Person):
     def __init__(self,name,age):
      super().__init__(name)
      self.age=age
student1 = Student("Abdullah", 20)

print(student1.name)
print(student1.age)
```

OVERRIDING

Is mein child class parent ke existing method ko apne tareeqe se redefine karti hai.

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):
    def sound(self):
        print("Dog barks")


dog1 = Dog()
dog1.sound()
```

```python
class Animal:
    def eat(self):
        print("Animal is eating")


class Dog(Animal):
    def eat(self):
        print("Dog is eating")


dog1 = Dog()

dog1.eat()
```

```python
class Person:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        print("My name is", self.name)


class Student(Person):
    def __init__(self, name, age):
        super().__init__(name)
        self.age = age

    def introduce(self):
        print("My name is", self.name, "and I am", self.age, "years old")

    def study(self):
        print("Student is studying")


student1 = Student("Abdullah", 20)

student1.introduce()
student1.study()
```
