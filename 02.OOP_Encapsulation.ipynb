Encapsulation

Encapsulation ka matlab hai class ke data ko protect/control karna aur direct access ko restrict karna.

Python mein commonly:

Public → name
Protected → _name
Private → __name

```python

class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Abdullah", 20)

print(student1.name)
print(student1.age)
```

```python
class Car:
    def __init__(self, brand,model):
        self.model = model
        self.brand = brand

car1 = Car("Haval", "Honda")

print(car1.brand)
print(car1.model)
```

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self._age = age

student1 = Student("Abdullah", 20)

print(student1.name)
print(student1._age)
```

```python
class BankAccount:
    def __init__(self, name, balance):
        self.name = name
        self._balance = balance

data = BankAccount("Hassan", 6660)

print(data.name)
print(data._balance)
```

Private Attribute

Double underscore __ attribute ko private banata hai.

```python
class BankAccount:
    def __init__(self, name, balance):
        self.name = name
        self._balance = balance

data = BankAccount("Hassan", 6660)

print(data.name)
print(data._balance)
```

Private Attribute ko Method se Access Karna

```python
class BankAccount:
    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance

    def get_balance(self):
        return self.__balance

account1 = BankAccount("Abdullah", 5000)

print(account1.get_balance())
```

```python
class BankAccount:
    def __init__(self, name, balance):
        self.name = name
        # balance ko private rakhne ke liye
        self.__balance = balance

    def get_balance(self):
        # method se private balance access karna
        return self.__balance


debt_account = BankAccount("Hassan", 500000)

print(debt_account.get_balance())
```

```python
class BankAccount:
    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def show_info(self):
        print("My name is", self.name, "and my balance is", self.__balance)


account1 = BankAccount("Abdullah", 10000)

print(account1.name)
print(account1.get_balance())
account1.show_info()
```
