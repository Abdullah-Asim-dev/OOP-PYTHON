__init__ Constructor

Abhi values class ke andar fixed hain:

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Abdullah", 20)
student2 = Student("Ali", 21)

print(student1.name)
print(student2.name)
```

```python
class Car:
    def __init__(self, model, brand):
        self.model = model
        self.brand = brand

car1 = Car("Civic", "Honda")
car2 = Car("Toyota", "GLI")

print(car1.model)
print(car2.model)
print(car1.brand)
print(car2.brand)
```

Important Points

__init__() → constructor

self → current object ko represent karta hai

self.model → object ka model

self.brand → object ka brand

car1 aur car2 → alag-alag objects

self ka basic idea ye hai ke jis object par kaam ho raha hai, self usi object ko refer karta hai.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Abdullah", 20)
student2 = Student("Ali", 21)
print(student1.name)
print(student1.age)
print(student2.name)
print(student2.age)
```

Methods

Ab object ke andar function banana seekhte hain. OOP mein class ke andar function ko method kehte hain.

```python
class Student:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        print("My name is", self.name)

student1 = Student("Abdullah")

student1.introduce()
```

```python
class Car:
    def __init__(self, model):
        self.model = model

    def info(self):
        print("My model is", self.model)

car1 = Car("Toyota")

car1.info()
```

```python
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def info(self):
        print(self.brand, self.model)

    def start(self):
        print("Car is starting")

car1 = Car("Honda", "Civic")

car1.info()
car1.start()
```

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(self.name, self.age)

    def study(self):
        print("Student is Starting ")

student1 = Student("Uman", 19)

student1.introduce()
student1.study()
```
